#### Outliers ko kese dhunde methods 

```python
plt.boxplot(data['Sepal_Length'])
plt.show
```

```python
plt.boxplot(data['Sepal_Width'])
plt.show
```

abhi outliers ki value replace krni hai remove nh krni hai 

 Apply the interquartile range(iqr) technique to identify and filter values outside the upper and lower boundaries.


```python
q3 = data['Sepal_Width'].quantile(0.75)
q1 = data['Sepal_Width'].quantile(0.25)
```

```python
iqr = q3 - q1
```

```python
lb = q1 - 1.5 * iqr
ub = q3 + 1.5 * iqr
```

```python
data[(data['Sepal_Width'] < lb) | (data['Sepal_Width'] > ub)]
```

how to handle :- using iloc

```python
data.loc[data['Sepal_Width'] < lb, 'Sepal_Width'] = lb
```

```python
plt.boxplot(data['Sepal_Width'])
plt.show()
```

```python
data.loc[data['Sepal_Width'] > ub, 'Sepal_Width'] = ub
```

```python
plt.boxplot(data['Sepal_Width'])
plt.show()
```

to replace null value 

```python
### Replace na value
median = data['Sepal_Width'].median()
data['Sepal_Width'].fillna(median)
```
