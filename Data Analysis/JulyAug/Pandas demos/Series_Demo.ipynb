## Series
1. pd.Series()
2. Homogeneous
3. Ordered
4. Mutable, Size-Immutable
5. Duplication allowed

```python
s1 = pd.Series([10, 20, 30, 40, 3.14])
s1 = pd.Series(['a', 'b', 'c'])
s1 = pd.Series([10, 20, 30, 40, 'a'])
s1
```

Customizable indexing

```python
# s2 = pd.Series(['a', 'b', 'c'])
# s2
```

```python
s1 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
s1
```
